# data Validation

```python
#check for null entries

for name, df in {
    "Users": users,
    "Products": products,
    "Orders": orders,
    "Order Items": order_items,
    "Reviews": reviews,
    "Events": events
}.items():

    print("\n" + "="*50)
    print(name)
    print(df.isnull().sum())
```

```python
#check duplicate rows

print(users.duplicated().sum())
print(products.duplicated().sum())
print(orders.duplicated().sum())
print(order_items.duplicated().sum())
print(reviews.duplicated().sum())
print(events.duplicated().sum())
```

```python
#check Primary Key is not duplicate

users["user_id"].duplicated().sum()

products["product_id"].duplicated().sum()

orders["order_id"].duplicated().sum()

order_items["order_item_id"].duplicated().sum()

reviews["review_id"].duplicated().sum()

events["event_id"].duplicated().sum()
```

```python
#check for item total is correct for each row

(order_items["quantity"] * order_items["item_price"] != order_items["item_total"]).sum()
```

```python
#visualize for mismatch

mismatch = order_items[
    order_items["quantity"] * order_items["item_price"] != order_items["item_total"]
]

mismatch.head(10)
```

```python
#check for round off matching prevents mismatch

(
    round(order_items["quantity"] * order_items["item_price"], 2)
    != round(order_items["item_total"], 2)
).sum()
```

```python
#check for values is closer to correct values 

import numpy as np

np.isclose(
    order_items["quantity"] * order_items["item_price"],
    order_items["item_total"]
).sum()
```

```python
#chrck for negative values

print("Negative Quantity:", (order_items["quantity"] < 0).sum())
print("Negative Item Price:", (order_items["item_price"] < 0).sum())
print("Negative Item Total:", (order_items["item_total"] < 0).sum())
print("Negative Product Price:", (products["price"] < 0).sum())
```
